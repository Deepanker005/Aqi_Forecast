**Apply FB Prophet Model to NewData here**

```python
from prophet import Prophet


model = Prophet()
model.fit(NewData)
```

```python
future = model.make_future_dataframe(periods=365, freq='h')
future.tail()
```

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()

```

```python
fig1 = model.plot(forecast)

```

```python
fig2 = model.plot_components(forecast)
```

### MultiVariate FB Prophet Analysis

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt

# 1. Prepare the Data
df = pd.DataFrame()
df['ds'] = NewData['ds'] 
df['y'] = AQI['C6H6(GT)']   # Target pollutant: Benzene
df['T'] = AQI['T']          # Regressor 1: Temperature
df['AH'] = AQI['AH']        # Regressor 2: Absolute Humidity
df.head()
```

```python
df.isnull().sum()
```

```python
# Chronological Train-Test Split (80% train, 20% test)
split_index = int(len(df) * 0.8)
train = df.iloc[:split_index]
test = df.iloc[split_index:]
```

```python
# Initialize and Train
model = Prophet(yearly_seasonality=True, weekly_seasonality=True, daily_seasonality=True)
model.add_regressor('T')
model.add_regressor('AH')
model.fit(train)
```

```python
# Predict on the Test Set
# The future dataframe must contain the timestamps AND regressor values
future = test[['ds', 'T', 'AH']]
forecast = model.predict(future)
```

```python
y_true = test['y'].values
y_pred = forecast['yhat'].values

rmse = np.sqrt(mean_squared_error(y_true, y_pred))
print(f"RMSE: {rmse:.2f} microg/m^3")
```

```python
fig = model.plot(forecast)
plt.show()
```

```python
fig2 = model.plot_components(forecast)
```
