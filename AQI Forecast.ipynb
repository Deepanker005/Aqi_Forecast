According to the dataset link all the null values are -200 thus we need to substitute those values with NAN so python can read it

```python
# Counting the -200 values in data

AQI.isin([-200]).sum(axis=0)
```

**Handling the Missing values**

```python
AQI=AQI.replace(to_replace=-200,value=np.nan)# Replacing -200 with nan
```

```python
AQI.isin([-200]).sum(axis=0)
AQI.isnull().sum()
```

```python
# Isolate the columns you want to calculate (excluding Date, Time, and ghost columns)
numeric_cols = []

# Loop through every single column name currently in the dataset
for col in AQI.columns:
    
    #Check our two conditions:
    #    - Is it NOT the 'Date' or 'Time' column?
    #    - Does the name NOT start with 'Unnamed'?
    if col not in ['Date', 'Time'] and not col.startswith('Unnamed'):
        
        # 4. If both conditions are true, add this column name to our list
        numeric_cols.append(col)

numeric_cols
```

```python
AQI[numeric_cols].mean()
```

```python
AQI[numeric_cols] = AQI[numeric_cols].fillna(AQI[numeric_cols].mean()) #Now calculate the mean of the DATA in those columns, and fill the NaNs
```

```python
AQI.isnull().sum()
```
